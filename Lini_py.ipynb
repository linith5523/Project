```python
%%writefile app.py

import streamlit as st
import numpy as np
import matplotlib.pyplot as plt

st.set_page_config(page_title="Conic Section Identifier")

st.title("Conic Section Identifier")
st.write("Identify and rotate a conic section using its quadratic equation.")

st.subheader("Enter the coefficients")

A = st.number_input("A", value=1.0)
B = st.number_input("B", value=0.0)
C = st.number_input("C", value=4.0)
D = st.number_input("D", value=0.0)
E = st.number_input("E", value=0.0)
F = st.number_input("F", value=-4.0)

if st.button("Analyze Conic"):

    # Quadratic matrix
    Q = np.array([
        [A, B / 2],
        [B / 2, C]
    ])

    # Eigenvalues and eigenvectors
    eigenvalues, eigenvectors = np.linalg.eigh(Q)

    # Determinant
    det_Q = np.linalg.det(Q)

    # Classification
    if abs(det_Q) < 1e-10:
        conic_type = "Parabola"
    elif det_Q > 0:
        conic_type = "Ellipse"
    else:
        conic_type = "Hyperbola"

    # Rotation angle
    theta = 0.5 * np.arctan2(B, A - C)
    angle = np.degrees(theta)

    # Equation
    equation = (
        f"{A}x² + {B}xy + {C}y² + "
        f"{D}x + {E}y + {F} = 0"
    )

    st.success(f"Conic Type: {conic_type}")

    st.subheader("Equation")
    st.write(equation)

    st.subheader("Quadratic Matrix Q")
    st.write(Q)

    st.subheader("Eigenvalues")
    st.write(eigenvalues)

    st.subheader("Eigenvectors")
    st.write(eigenvectors)

    st.subheader("Rotation")
    st.write(f"Rotation angle: {angle:.2f}°")

    # Graph
    x = np.linspace(-5, 5, 500)
    y = np.linspace(-5, 5, 500)

    X, Y = np.meshgrid(x, y)

    Z = (
        A * X**2
        + B * X * Y
        + C * Y**2
        + D * X
        + E * Y
        + F
    )

    fig, ax = plt.subplots()

    ax.contour(X, Y, Z, levels=[0])

    ax.axhline(0)
    ax.axvline(0)

    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_title(f"{conic_type} — Rotation {angle:.2f}°")

    ax.grid(True)

    st.pyplot(fig)
```
